## Step 1 · A local random generator

Create a generator with `np.random.default_rng(seed)`. A seed makes the same sequence reproducible for the same generator setup and calls; it does not make a sample equal to its theoretical distribution. Calling the generator advances its state. Avoid relying on exact random values across arbitrary future NumPy versions.

```python
rng = np.random.default_rng(42)
print("Dice:", rng.integers(1, 7, size=12))
print("Uniform:", rng.uniform(0, 1, size=4))
print("Normal:", rng.normal(loc=170, scale=10, size=4))
a = np.random.default_rng(123).integers(0, 100, 5)
b = np.random.default_rng(123).integers(0, 100, 5)
np.testing.assert_array_equal(a, b)
```

## Step 2 · Monte Carlo probability

Monte Carlo estimates a quantity by repeated random trials. Each row below is one trial with two dice. A Boolean event is 1 for success and 0 for failure, so its mean estimates probability. The standard error describes typical sampling variation; it is not a guarantee for a particular run.

```python
rng = np.random.default_rng(7)
dice = rng.integers(1, 7, size=(50_000, 2))
event = dice.sum(axis=1) == 7
estimate = event.mean()
standard_error = np.sqrt(estimate * (1 - estimate) / event.size)
print("Estimated P(sum=7):", estimate)
print("Exact probability:", 6 / 36)
print("Estimated standard error:", standard_error)
assert 0 <= estimate <= 1
```

## Step 3 · Bootstrap a sample mean

A bootstrap resamples the observed dataset with replacement. Each row is a new sample of the same size. Quantiles of bootstrap means give a simple percentile interval. This illustration assumes independent, representative observations; resampling cannot repair biased data.

```python
observed = np.array([12., 14., 9., 15., 13., 11., 18., 16.])
rng = np.random.default_rng(10)
resampled = rng.choice(observed, size=(2_000, observed.size), replace=True)
bootstrap_means = resampled.mean(axis=1)
interval = np.quantile(bootstrap_means, [0.025, 0.975])
print("Observed mean:", observed.mean())
print("Illustrative 95% percentile interval:", interval)
assert bootstrap_means.shape == (2_000,)
```

### Solution 1

Generate 100 die rolls (integers 1 through 6) with a generator seeded with 99.

**Why it works:** integers excludes its high bound, so use 7 to allow 6. A fresh generator with seed 99 reproduces the requested sequence.

```python
reference_rolls = np.random.default_rng(99).integers(1, 7, size=100)
assert reference_rolls.shape == (100,)
assert np.issubdtype(reference_rolls.dtype, np.integer)
assert np.all((reference_rolls >= 1) & (reference_rolls <= 6))
np.testing.assert_array_equal(reference_rolls, np.random.default_rng(99).integers(1, 7, size=100))
print(reference_rolls)
```

### Solution 2
