## Step 1 · Indexing and slicing

Use `a[row, column]` for a 2D array. A slice is `start:stop:step`; stop is excluded. A missing start or stop means the boundary. Negative indices count from the end. A single integer removes an axis; a slice preserves it.

```python
scores = np.array([[65, 80, 90], [75, 60, 85], [95, 88, 92], [50, 70, 68]])
print("Last student's last score:", scores[-1, -1])
print("First two students:\n", scores[:2])
print("Second subject:", scores[:, 1])
print("Shapes:", scores[:, 1].shape, scores[:, 1:2].shape)
print("Every other student:\n", scores[::2])
```

**Try it:** Select the last two students and the first two subjects.

## Step 2 · Boolean masks

A comparison creates a Boolean array. A mask chooses entries where it is True. Combine array conditions with `&` (and), `|` (or), and `~` (not), and put each comparison in parentheses. Python `and` cannot combine elementwise array conditions.

```python
marks = np.array([45, 62, 78, 90, 55, 84])
mask = (marks >= 60) & (marks < 85)
print("Mask:", mask)
print("Selected:", marks[mask])
print("Positions:", np.flatnonzero(mask))
assert np.count_nonzero(mask) == 3
```

**Try it:** Select marks below 50 or at least 85.

## Step 3 · Choose rows and columns carefully

Two integer index arrays select paired coordinates. To select every combination of some rows and columns, use `np.ix_`. Assign through one indexing expression to change the original array. Later we will study copies and views in detail.

```python
table = np.arange(12).reshape(3, 4)
print(table)
print("Paired:", table[[0, 2], [1, 3]])
print("Rectangle:\n", table[np.ix_([0, 2], [1, 3])])
adjusted = table.copy()
adjusted[adjusted < 4] = -1
print("Adjusted:\n", adjusted)
```

**Try it:** Select the middle row in reverse order.

### Solution 1

Select all rows and the last two columns of grid.

**Why it works:** The first colon keeps every row. The -2: slice selects the final two columns and preserves both axes.

```python
reference_last_columns = grid[:, -2:]
np.testing.assert_array_equal(reference_last_columns, [[3, 4], [8, 9], [13, 14], [18, 19]])
print(reference_last_columns)
```

### Solution 2

Keep prices that are at least 100 and at most 300.

**Why it works:** Each comparison creates a mask. Their elementwise intersection keeps both endpoints and rejects values outside the interval.

```python
reference_affordable = prices[(prices >= 100) & (prices <= 300)]
np.testing.assert_array_equal(reference_affordable, [100, 150, 300])
print(reference_affordable)
```
