## Step 2 · Lists, loops, and functions

A list stores a sequence. Indexing begins at zero. A `for` loop repeats work. A function gives a reusable name to a calculation; `return` sends its result back. Indentation is part of Python syntax.

```python
prices = [10, 20, 30]
print("First price:", prices[0])
discounted = [price * 0.9 for price in prices]
def total_with_tax(values, tax_rate=0.05):
    return sum(values) * (1 + tax_rate)
print(discounted)
print(total_with_tax(prices))
```

**Try it:** Call the function with tax_rate=0.10. Explain what changed.

## Step 3 · Meet NumPy and read an error

NumPy provides arrays for numerical calculations. The usual import name is `np`. Here, multiplying an array changes every value. Multiplying a Python list repeats its contents. Errors are learning tools: read the last line first. We catch this intentional error so Run All still works.

```python
values = np.array([10, 20, 30])
print("List:", [10, 20, 30] * 2)
print("Array:", values * 2)
try:
    print(values[99])
except IndexError as error:
    print("Intentional indexing error:", error)
np.testing.assert_array_equal(values * 2, [20, 40, 60])
```

**Try it:** Replace index 99 with 2. What is the largest valid positive index?
