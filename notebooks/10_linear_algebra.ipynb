## Step 1 · Matrix multiplication combines weighted values

For A shaped (m, n) and B shaped (n, p), A @ B has shape (m, p). Each result is a row-column dot product: multiply matching entries and add them. A matrix times a length-n vector produces a length-m vector.

```python
A = np.array([[1., 2.], [3., 4.]])
v = np.array([10., 20.])
print("Elementwise:\n", A * A)
print("Matrix product:\n", A @ A)
print("Matrix times vector:", A @ v)
np.testing.assert_allclose(A @ v, [50, 110])
```

## Step 2 · Solve equations and check residuals

To solve A x = b for square, full-rank A, use `linalg.solve`. Computing an explicit inverse just to multiply by b usually adds work and numerical error. A residual A @ x - b checks how closely the computed solution satisfies the equations. Nearly dependent columns can make solutions sensitive to input changes.

```python
A = np.array([[2., 1.], [1., 3.]])
b = np.array([5., 7.])
x = np.linalg.solve(A, b)
print("Solution:", x)
print("Residual norm:", np.linalg.norm(A @ x - b))
print("Condition number:", np.linalg.cond(A))
np.testing.assert_allclose(A @ x, b)
```

**Try it:** Make both rows identical and catch the resulting LinAlgError.

## Step 3 · Fit a line and factor a matrix

For more equations than unknowns, least squares minimizes the sum of squared residuals. A column of ones represents the intercept. SVD factors a matrix into orthogonal directions and singular values; small singular values identify directions with little variation. We reconstruct the design matrix to check the factorization.

```python
x_data = np.arange(5, dtype=float)
y_data = np.array([1.1, 2.9, 5.2, 6.8, 9.0])
design = np.column_stack([np.ones(x_data.size), x_data])
coefficients, residuals, rank, singular_values = np.linalg.lstsq(design, y_data, rcond=None)
prediction = design @ coefficients
print("Intercept, slope:", coefficients)
print("RMSE:", np.sqrt(np.mean((prediction - y_data) ** 2)))
U, s, Vt = np.linalg.svd(design, full_matrices=False)
np.testing.assert_allclose((U * s) @ Vt, design, atol=1e-12)
assert rank == 2
```
