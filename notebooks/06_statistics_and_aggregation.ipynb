## Step 1 · Mean, median, and spread

The mean is sensitive to extreme values; the median is the middle ordered value. Standard deviation measures spread around the mean. NumPy defaults to `ddof=0`, dividing variance by N. `ddof=1` divides by N-1 and is commonly used to estimate population variance from a sample with more than one value.

```python
delivery_minutes = np.array([20., 22., 21., 23., 90.])
print("Mean:", delivery_minutes.mean())
print("Median:", np.median(delivery_minutes))
print("Population std:", delivery_minutes.std(ddof=0))
print("Sample std:", delivery_minutes.std(ddof=1))
print("Quartiles:", np.percentile(delivery_minutes, [25, 50, 75]))
```

## Step 2 · Reduce the intended axis

Use a 2D table to separate questions: totals per day reduce products, and totals per product reduce days. `argmax` returns a position, while `max` returns a value. We use both to label the largest result.

```python
sales = np.array([[5, 3, 2], [7, 2, 4], [4, 6, 3], [8, 4, 5]])
products = np.array(["tea", "coffee", "juice"])
daily = sales.sum(axis=1)
product_totals = sales.sum(axis=0)
print("Daily:", daily)
print("Product totals:", product_totals)
print("Top product:", products[product_totals.argmax()])
assert daily.sum() == product_totals.sum()
```

**Try it:** Find the best day for each product using argmax(axis=0).

## Step 3 · Weighted and cumulative summaries

A weighted mean lets observations contribute differently; weights must have a nonzero sum. Cumulative sums track running totals. Differences compare adjacent observations and shorten that axis by one.

```python
marks = np.array([70., 80., 90.])
weights = np.array([0.2, 0.3, 0.5])
print("Weighted grade:", np.average(marks, weights=weights))
daily_sales = np.array([100., 150., 120., 180.])
print("Running total:", np.cumsum(daily_sales))
print("Changes:", np.diff(daily_sales))
np.testing.assert_allclose(np.average(marks, weights=weights), 83.)
```

**Try it:** Compute the running mean with cumsum divided by arange(1, n+1).

### Solution 2

Compute the running mean of spending, including the current day.

**Why it works:** The cumulative sum contains the total so far. Divide each entry by the number of days included at that position.

```python
reference_running_mean = np.cumsum(spending) / np.arange(1, spending.size + 1)
np.testing.assert_allclose(reference_running_mean, [10, 20, 20, 25])
print(reference_running_mean)
```
