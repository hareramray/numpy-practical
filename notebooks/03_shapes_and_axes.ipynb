## Step 1 · Reshape preserves the number of values

A shape is a contract about how values are arranged. `reshape` does not add or remove values. One dimension may be -1: NumPy calculates its length from the others. A 1D vector has no separate row and column axes; use a new axis when you need a column.

```python
values = np.arange(12)
table = values.reshape(3, 4)
print(table)
print("Infer columns:\n", values.reshape(2, -1))
print("Column shape:", values[:, None].shape)
print("Row shape:", values[None, :].shape)
assert table.size == values.size
```

## Step 2 · Transpose and reductions

Transpose rearranges axes. For a 2D table, `.T` exchanges rows and columns. A reduction collapses the axis you name: for shape (students, subjects), summing axis 0 gives one result per subject. Summing axis 1 gives one result per student.

```python
points = np.array([[1, 2, 3], [4, 5, 6]])
print("Transposed:\n", points.T)
print("Per-column totals:", points.sum(axis=0))
print("Per-row totals:", points.sum(axis=1))
print("Keep the reduced axis:", points.sum(axis=1, keepdims=True).shape)
np.testing.assert_array_equal(points.sum(axis=0), [5, 7, 9])
```

## Step 3 · Concatenate versus stack

`concatenate` joins along an existing axis, so other axis lengths must match. `stack` adds a new axis and requires equal input shapes. `array_split` can divide into uneven pieces; `split` requires equal pieces when given an integer count.

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print("Concatenate:", np.concatenate([a, b]))
print("Stack rows:\n", np.stack([a, b], axis=0))
print("Stack columns:\n", np.stack([a, b], axis=1))
print("Uneven pieces:", np.array_split(np.arange(7), 3))
```
