## Step 1 · Elementwise calculations

Operations such as +, -, *, / and ** apply to corresponding entries. Universal functions (ufuncs) such as `sqrt`, `exp`, and `log` also operate elementwise. `*` is elementwise multiplication; matrix multiplication uses `@` in a later notebook.

```python
quantity = np.array([2, 3, 1])
price = np.array([50., 20., 100.])
revenue = quantity * price
print("Revenue by product:", revenue)
print("Total:", revenue.sum())
print("Square roots:", np.sqrt([1., 4., 9.]))
np.testing.assert_allclose(np.log(np.exp([0., 1., 2.])), [0, 1, 2])
```

## Step 2 · Conditions and clipping

`where(condition, yes, no)` chooses an output per element. `clip` limits values to an interval. The arguments to `where` are evaluated before selection; it does not prevent an invalid calculation inside an argument.

```python
scores = np.array([-5, 45, 75, 110])
bounded = np.clip(scores, 0, 100)
labels = np.where(bounded >= 60, "pass", "retry")
print(bounded)
print(labels)
np.testing.assert_array_equal(bounded, [0, 45, 75, 100])
```

## Step 3 · Safe division with an output buffer

Use a ufunc's `where=` argument to avoid calculating invalid entries. Initialize `out` first so masked entries have a known value. Here an undefined unit price is represented by NaN, a floating point missing-value marker.

```python
revenue = np.array([100., 0., 90.])
units = np.array([2., 0., 3.])
unit_price = np.full_like(revenue, np.nan)
np.divide(revenue, units, out=unit_price, where=units != 0)
print(unit_price)
np.testing.assert_allclose(unit_price, [50, np.nan, 30], equal_nan=True)
```
