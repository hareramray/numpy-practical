## Step 1 · Read the shape

An array is a rectangular collection of values with one dtype. `shape` gives the length of each axis, `ndim` counts axes, and `size` counts all elements. For a table, axis 0 runs through rows and axis 1 through columns.

```python
sales = np.array([[12, 15, 18], [20, 22, 25]], dtype=np.int64)
print(sales)
print("shape:", sales.shape, "ndim:", sales.ndim, "size:", sales.size)
print("dtype:", sales.dtype, "bytes:", sales.nbytes)
assert sales.shape == (2, 3)
assert sales.size == 6
```

## Step 2 · Choose a constructor

`arange(start, stop, step)` excludes stop. `linspace(start, stop, num)` includes both endpoints by default and is convenient for floating point grids. `zeros`, `ones`, and `full` create known initial values. `empty` allocates uninitialized numerical storage: fill every entry before reading it.

```python
print("Even numbers:", np.arange(0, 10, 2))
print("Five points:", np.linspace(0, 1, 5))
print("Zeros:\n", np.zeros((2, 3)))
print("Full:\n", np.full((2, 2), 7))
buffer = np.empty(3)
buffer[:] = 4.5
print("Filled buffer:", buffer)
```

**Try it:** Create seven evenly spaced points from -3 to 3.

## Step 3 · Dtype affects results

Integers cannot store a fractional part. Converting floats to integers truncates toward zero. Small integer types have limited ranges; inspect `iinfo` before choosing one. Use explicit floating point arrays when your calculation needs fractions.

```python
prices = np.array([1, 2, 3], dtype=np.int64)
prices[0] = 1.9
print("After integer assignment:", prices)
floating = prices.astype(np.float64)
floating[0] = 1.9
print("Floating version:", floating)
print("int8 limits:", np.iinfo(np.int8).min, np.iinfo(np.int8).max)
assert floating.dtype == np.float64
```

**Try it:** Compare nbytes for 1,000 values stored as float32 and float64.

### Solution 1

Create a 3-row, 4-column array filled with 2.5, using float64.

**Why it works:** full takes the requested shape and fill value. Explicit float64 preserves the fractional part.

```python
reference_filled = np.full((3, 4), 2.5, dtype=np.float64)
assert reference_filled.shape == (3, 4)
assert reference_filled.dtype == np.float64
np.testing.assert_allclose(reference_filled, 2.5)
print(reference_filled)
```

### Solution 2

Create an integer array containing 5, 10, 15, 20, 25.

**Why it works:** arange excludes its stop, so 26 includes 25 when counting from 5 in steps of 5.

```python
reference_multiples = np.arange(5, 26, 5)
np.testing.assert_array_equal(reference_multiples, [5, 10, 15, 20, 25])
assert np.issubdtype(reference_multiples.dtype, np.integer)
print(reference_multiples)
```
