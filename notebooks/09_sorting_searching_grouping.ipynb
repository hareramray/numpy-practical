## Step 1 · Sort values and positions

`sort` returns sorted values. `argsort` returns indices that reorder the original. Reuse those indices for related columns so names stay aligned with scores. Stable sorting preserves the input order of ties; reversing an ascending result reverses ties too, so sort negative scores for stable descending order.

```python
names = np.array(["Asha", "Ben", "Chen", "Divya"])
scores = np.array([82, 95, 82, 70])
order = np.argsort(-scores, kind="stable")
print(names[order])
print(scores[order])
np.testing.assert_array_equal(names[order], ["Ben", "Asha", "Chen", "Divya"])
```

## Step 2 · Search in sorted data

`searchsorted` returns insertion positions in an ascending sorted array. `side="left"` inserts before equal values; `side="right"` inserts after them. Their difference counts duplicates. `unique` finds distinct labels and can return group indices and counts.

```python
sorted_values = np.array([10, 20, 20, 30, 40])
left = np.searchsorted(sorted_values, 20, side="left")
right = np.searchsorted(sorted_values, 20, side="right")
print("Insertion range:", left, right, "count:", right - left)
labels, counts = np.unique(["tea", "coffee", "tea", "juice"], return_counts=True)
print(dict(zip(labels, counts)))
assert right - left == 2
```

## Step 3 · Group totals without pandas

`unique(..., return_inverse=True)` encodes each label as a nonnegative integer. `bincount` adds weights by integer code. This gives group sums; dividing by counts gives group means. Weighted bincount returns floating point output.

```python
category = np.array(["tea", "coffee", "tea", "juice", "coffee"])
revenue = np.array([10., 20., 15., 12., 30.])
groups, inverse = np.unique(category, return_inverse=True)
totals = np.bincount(inverse, weights=revenue, minlength=groups.size)
counts = np.bincount(inverse, minlength=groups.size)
for group, total, mean in zip(groups, totals, totals / counts):
    print(group, "total:", total, "mean:", mean)
np.testing.assert_allclose(totals.sum(), revenue.sum())
```

**Try it:** Compute the share of total revenue contributed by each group.
